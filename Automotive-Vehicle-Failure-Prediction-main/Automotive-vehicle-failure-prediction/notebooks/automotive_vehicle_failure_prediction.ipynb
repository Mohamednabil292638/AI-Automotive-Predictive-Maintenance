## Data Quality Investigation — Suspicious Brake Temperature Values

During the initial data-quality assessment, the `brake_temp_c` variable was found to contain 94 negative observations.

These values were flagged as suspicious because brake temperature is substantially below the corresponding ambient temperature in several observations.

For example:

- `brake_temp_c = -0.58°C` while `ambient_temp_c = 16.04°C`
- `brake_temp_c = -4.21°C` while `ambient_temp_c = 34.99°C`
- `brake_temp_c = -25.83°C` while `ambient_temp_c = 12.08°C`
- `brake_temp_c = -31.86°C` while `ambient_temp_c = 36.99°C`
- `brake_temp_c = -45.55°C` while `ambient_temp_c = 25.67°C`

The surrounding telemetry variables, including vehicle speed, brake pedal position and brake-pad wear, were examined to provide operating context.

These observations are therefore treated as **potential data-quality anomalies** rather than being removed immediately.

### Initial conclusion

The negative brake-temperature values require further investigation before deciding whether they represent:

1. Sensor/data-generation errors,
2. Measurement offsets,
3. Genuine operating conditions, or
4. Another systematic data-quality issue.

No records are removed at this stage.

## Investigation Findings — Suspicious Sensor Values

### 1. Brake Temperature

The `brake_temp_c` variable contains 94 negative observations.

A summary of these observations shows:

- Mean brake temperature: -22.12°C
- Minimum brake temperature: -84.88°C
- Maximum brake temperature: -0.07°C
- Mean ambient temperature: 29.62°C
- Mean vehicle speed: 63.22 km/h
- Mean brake pedal position: 45.86%

The negative brake temperatures occur while the corresponding ambient temperatures are positive. Several observations also show substantial vehicle speed and brake pedal activity.

A correlation analysis within these suspicious records showed:

- Brake temperature vs ambient temperature: -0.073
- Brake temperature vs vehicle speed: -0.249
- Brake temperature vs brake pedal position: 0.074

These relationships are weak and do not by themselves establish whether the sensor values are incorrect.

### Finding

The negative brake-temperature observations appear to represent a potential systematic data-quality issue rather than isolated outliers. Further investigation is required before deciding whether these values should be removed, transformed, or retained as anomalies.

---

### 2. Vibration Level

One observation contains a negative `vibration_level`:

- Vehicle: VEH0008
- Vibration level: -0.004329
- Engine RPM: 2460.83
- Engine load: 20.11%
- Vehicle speed: 61.47 km/h

The negative value is extremely close to zero and occurs during otherwise normal vehicle operation.

### Finding

This observation is treated as a minor potential measurement or rounding anomaly. It does not currently provide sufficient evidence for removing the observation.

### Data Quality Decision

At this stage, suspicious observations are identified but not automatically deleted. Treatment will be decided after understanding the data-generating process and the requirements of the downstream analysis.

## Data Cleaning Decision Framework

Abnormal values should not be treated solely according to their magnitude.

The treatment depends on the meaning of the variable, the magnitude of the anomaly, the surrounding observations, and the likely cause of the issue.

### Vibration Level

A single observation contained a very small negative vibration value (`-0.004329`). Since vibration magnitude is expected to be non-negative and the deviation is extremely small, this may represent measurement or numerical noise.

**Potential treatment:** convert the value to `0`, subject to validation.

### Brake Temperature

Ninety-four observations contained negative brake temperatures, ranging from approximately `-0.07°C` to `-84.88°C`.

Because these observations appear to represent a systematic issue rather than an isolated small deviation, they should not simply be converted to positive values.

**Potential treatment:** flag the observations and investigate the data-generation process. If confirmed as invalid measurements, the affected `brake_temp_c` values may be converted to missing values while retaining the remaining vehicle telemetry.

### General Principle

Anomalous values should be investigated before modification. Values should never be changed to an assumed positive value without evidence that the transformation represents the true measurement.

## Statistical Question — Engine Overheat

### Business Question
Is engine temperature associated with Engine Overheat observations?

### Statistical Question
Is there a statistically significant difference in engine temperature between
No Failure observations and Engine Overheat observations?

### Variable
- Numerical variable: `engine_temp_c`

### Groups
- Group 1: `No Failure`
- Group 2: `Engine Overheat`

### Hypotheses

**Null hypothesis (H₀):**
There is no difference in the mean engine temperature between the two groups.

**Alternative hypothesis (H₁):**
There is a difference in the mean engine temperature between the two groups.

### Important Limitation
The dataset contains only 4 observations classified as `Engine Overheat`,
compared with 1,935 `No Failure` observations. Therefore, any statistical
conclusion should be treated as exploratory and interpreted cautiously.

A difference in averages does not by itself prove that engine temperature
causes engine overheating.

### Initial Statistical Observation

The Engine Overheat observations show substantially higher engine
temperatures than the No Failure observations.

- No Failure mean: 95.04°C
- Engine Overheat mean: 112.41°C
- Difference in means: approximately 17.37°C

All four Engine Overheat observations are above the 75th percentile
of the No Failure group (98.59°C).

This provides strong descriptive evidence of an association between
high engine temperature and Engine Overheat observations.

However, the Engine Overheat group contains only four observations.
Therefore, the result has a severe small-sample limitation and should
not be interpreted as proof that high engine temperature causes
engine overheating.

A statistical test can be used as exploratory evidence, but the
result must be interpreted together with the sample size and
engineering context.

```python
X = model_df.drop(columns=["failure_flag"])
y = model_df["failure_flag"]
```

```python
print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
categorical_columns=X.select_dtypes(include=["object"]).columns.tolist()
numerical_columns=X.select_dtypes(include=["int64","float64"]).columns.tolist()
print(categorical_columns)
print(numerical_columns)
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.20,random_state=33,stratify=y)
```

```python
print("Training set:")
print(y_train.value_counts())

print("\nTest set:")
print(y_test.value_counts())
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

preprocessor = ColumnTransformer(
    transformers=[
        (
            "num",
            StandardScaler(),
            numerical_columns
        ),
        (
            "cat",
            OneHotEncoder(handle_unknown="ignore"),
            categorical_columns
        )
    ]
)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression

logistic_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("classifier", LogisticRegression(
            class_weight="balanced",
            max_iter=1000,
            random_state=42
        ))
    ]
)
```

```python
logistic_model.fit(X_train, y_train)
```

```python
y_pred=logistic_model.predict(X_test)
```

```python
y_prob = logistic_model.predict_proba(X_test)

print(y_prob[:10])
```

```python
y_prob[:10]
```

```python
failure_probability = logistic_model.predict_proba(X_test)[:, 1]

print(failure_probability[:10])
```

```python
print(y_pred)
```

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_test, y_pred)

print(cm)
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_test, y_pred, zero_division=0))
```

```python
from sklearn.metrics import average_precision_score, roc_auc_score

print("PR-AUC:", average_precision_score(y_test, failure_probability))
print("ROC-AUC:", roc_auc_score(y_test, failure_probability))
```

## Logistic Regression — Baseline Results

The baseline Logistic Regression model was evaluated on a stratified
20% hold-out test set.

### Results

| Metric | Failure Class |
|---|---:|
| Precision | 0.88 |
| Recall | 1.00 |
| F1-score | 0.93 |

Additional ranking metrics:

- ROC-AUC: 0.9989
- PR-AUC: 0.9379

### Confusion Matrix

- True Negatives: 386
- False Positives: 1
- False Negatives: 0
- True Positives: 7

The model detected all 7 failure observations in the test set and
produced one false positive.

### Interpretation

The baseline model shows strong separation between failure and
no-failure observations on this particular test split.

However, the dataset is highly imbalanced, with only 35 failure
observations in the complete dataset and only 7 failures in the test
set. Therefore, these performance estimates are subject to high
uncertainty and should not be interpreted as evidence of production
readiness.

Furthermore, the current target represents failure status on the
observed record. Therefore, this model should be described as a
failure-event classification model rather than a validated future
failure prediction system.
